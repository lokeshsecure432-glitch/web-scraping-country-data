# Country Data Web Scraping Project
## Objective
The objective of this project is to extract country data such as country name, population, and area from a website using Python web scraping libraries.

## Import Required Libraries

```python
from bs4 import BeautifulSoup
import requests
import pandas as pd
```

## Send Request to Website

```python
url='https://www.scrapethissite.com/pages/simple/'
res=requests.get(url) 
if res.status_code==200:
    with open('countries_data.html','w',encoding='utf-8') as f:
        f.write(res.text)
else:
    print('data not fetch properly retry to fetch the data')
```

## Read and store HTML content from file into a variable.

```python
with open('countries_data.html','r') as f:
    html_content=f.read()
```

## Parse HTML Content Using BeautifulSoup

```python
soup=BeautifulSoup(html_content,'lxml') 
```

## Extract Country Data
### All the required information is stored near the h3 tags in the webpage structure.

```python
all_h3=soup.find_all("h3")
country_data=[]  
for h3 in all_h3:
    names=h3.get_text(strip=True)

    population=h3.find_next('div').select_one('span.country-population').get_text(strip=True)

    area=h3.find_next('div').select_one('span.country-area').get_text(strip=True)

    country_data.append([names,population,area])

```

## Store Extracted Data into Pandas DataFrame

```python
df=pd.DataFrame(country_data, columns=['country_name','population','area(km^2)'])
```

## Display Extracted Data

```python
df.head()
df.isnull().sum()
```

## Export Data to CSV File

```python
df.to_csv("country_data.csv", index=False)
```

## Conclusion
This project successfully extracted country data from a website using Python web scraping techniques and stored the data in CSV format for further analysis.
